```python
def parni(lista):
    return {
        'Parni': [x for x in lista if x % 2 == 0],
        'Neparni': [x for x in lista if x % 2 != 0]
    }
print(parni([1, 7, 2, 4, 5]))
```

```python
def numlista(lista):
    res = {}
    for x in lista:
        tip = type(x).__name__
        if tip in res:
            res[tip].append(x)
        else:
            res[tip] = [x]
    return res
print(numlista(["Prvi", "Drugi", 2, 4, [3, 5]]))
```

```python
def uredi(lista, n, x):
    res = []
    for i in range(len(lista)):
        if i < n:
            res.append(lista[i] + x)
        else:
            res.append(lista[i] - x)
    return res
print(uredi([1, 2, 3, 4, 5], 3, 1))
```

```python
def zbir(lista):
    return [lista[i] + lista[i + 1] for i in range(len(lista) - 1)]
print(zbir([1, 2, 3, 4, 5]))
```

```python
def brojel(*a):
    res = []
    for x in a:
        if (type(x) != list):
            res.append(-1)
        else:
            res.append(len(x))
    return res
print(brojel([1, 2], [3, 4, 5], 'el', ['1', 1]));
```

```python
def razlika(lista1, lista2):
    return [x for x in lista1 if x not in lista2]
print(razlika([1, 4, 6, "2", "6"], [4, 5, "2"]))
```

```python
def saberi(lista):
    return [sum(x) for x in lista]
print(saberi([(1, 4, 6), (2, 4), (4, 1)]))
```

```python
def izmeni(lista):
    return [sum(lista[0:i + 1]) for i in range(len(lista))]
print(izmeni([1, 2, 4, 7, 9]))
```

```python
def prosek(lista):
    return [sum(x) / len(x) for x in lista]
print(prosek([[1, 4, 6, 2], [4, 6, 2, 7], [3, 5], [5, 6, 2, 7]]))
```

```python
def izbroj(lista):
    res = 0
    for x in lista:
        if type(x) == list:
            res += izbroj(x)
        else:
            res += 1
    return res
print(izbroj([1, [1, 3, [2, 4, 5, [5, 5], 4]], [2, 4], 4, 6]))
```

```python
def razlika(lista):
    return [lista[i] - lista[i + 1] for i in range(len(lista) - 1)]
print(razlika([8, 5, 3, 1, 1]))
```

```python
def presek(lista1, lista2):
    return [x for x in lista1 if x in lista2]
print(presek([5, 4, "1", "8", 3, 7], [1, 9, "1"]))
```

```python
def izmeni(lista):
    return {
        'pp': [lista[i] + 1 for i in range(len(lista) - 1) if i % 2 == 0],
        'np': [lista[i] - 1 for i in range(len(lista) - 1) if i % 2 != 0]
    }
print(izmeni([8, 6, 3, 1, 1]))
```

```python
def unija(lista1, lista2):
    res = []
    for x in lista1 + lista2:
        if x not in res:
            res.append(x)
    return res
print(unija([5, 4, "1", "8", 7], [1, 9, "1"]))
```

```python
def izdvoji(*a):
    n = 0
    res = []
    for lista in a:
        res.append(lista[n] if len(lista) > n else 0)
        n += 1
    return res
print(izdvoji([5, 4, 4], [1, 9, 1], [5, 6], [4, 6, 10, 12]))
```

```python
def brojanje(d):
    count = {}
    for x in d.values():
        tip = type(x).__name__
        count[tip] = count.get(tip, 0) + 1
    return list(count.items())
print(brojanje({1 : 4, 2 : [2, 3], 3 : [5, 6], 4 : 'test', 5 : 9, 6 : 8}))
```

```python
def kreiraj(n):
    return [(i, sum(range(i + 1)) ** 2) for i in range(n + 1)]
print(kreiraj(4))
```

```python
def kreiraj(lista):
    return [[x for x in lista[i] if x not in lista[i + 1]] for i in range(len(lista) - 1)]
print(kreiraj([[1, 2, 3], [2, 4, 5], [4, 5, 6, 7], [1, 5]]))
```

```python
from functools import reduce
def stepenuj(lista):
    return [reduce(lambda x, y: x ** y, t) for t in lista]
print(stepenuj([(1, 4, 2), (2, 5, 1), (2, 2, 2, 2), (5, )]))
```

```python
def boje(hexa):
    num = hexa.lstrip('#')
    return {
        "Red": int(num[0:2], 16),
        "Green": int(num[2:4], 16),
        "Blue": int(num[4:6], 16)
    }
print(boje('#FA1AA0'))
```
